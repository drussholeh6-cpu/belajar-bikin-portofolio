```python
def fpb(a, b):
    for i in range(min(a, b), 0, -1):
        if a % i == 0 and b % i == 0:
            return i
```

```python
fpb(36,79)
```

```python
def fpb(a, b):
    faktor_sama = []
    for i in faktor(a):
        if i in faktor(b):
            faktor_sama.append(i)
    return max(faktor_sama)
```

```python
fpb(12,18)
```

```python
def relativeprima (a,b):
  if fpb(a,b)==1:
    print(a,'dan',b,'relative prima')
  else:
    print(a,'dan',b,'bukan relative prima')
```

```python
def kpk(a,b):
  kelipatan_a = []
  kelipatan_b = []
  for i in range(1,b+1):
    kelipatan_a.append(a*i)
  for j in range(1,a+1):
    kelipatan_b.append(b*j)
  sama = []
  for x in kelipatan_a:
    for y in kelipatan_b:
      if x==y:
        sama.append(x)
  return min(sama)
```

```python
kpk(7,8)
```

```python
def barprima(n):
    bprima = []
    x = 2  # Inisialisasi nilai awal x
    while len(bprima) < n:
        if prima(x) == True:
            bprima.append(x)
        x = x + 1
    return bprima  # Mengembalikan list hasil
```

```python
import sympy as sp
t = sp.symbols('t')
y = sp.Function('y')

pdb = sp.Eq((y(t)).diff(t,2) + 2*y(t).diff(t) + 5*y(t),0)
sol = sp.dsolve(pdb, y(t), ics={y(0): 1, y(t).diff(t).subs(t,0): 0})
sp.simplify(sol)
```

```python
s = sol.rhs
sp.simplify(s.diff(t, 2) + 2*s.diff(t) + 5*s)
```

```python
f, g = sp.Function("f"), sp.Function("g")

sistem = [sp.Eq(f(t).diff(t), f(t) + 2*g(t)),
          sp.Eq(g(t).diff(t), 3*f(t) + 2*g(t))]

[sp.simplify(h) for h in sp.dsolve(sistem)]
```

```python
import pandas as pd

# Deklarasi Series secara bertahap (data populasi, indeks/nama kota, dan nama series)
data_populasi = [909976, 8615246, 2874692, 4625649]
indeks_kota = ['Stockholm', 'London', 'Rome', 'Barcelona']

s = pd.Series(data_populasi, index=indeks_kota)
s.name = 'population'
print(s)

# Deklarasi Series secara lengkap sekaligus dalam satu perintah
s_pop = pd.Series(
    [909976, 8615246, 2874692, 4625649],
    index=['Stockholm', 'London', 'Rome', 'Barcelona'],
    name='population'
)
print(s_pop)

# Mengakses data berdasarkan indeks tertentu
print(s['London'])
print(s['Stockholm'])
```

```python
# Mencari nilai median, mean (rata-rata), dan standar deviasi
print("Median:", s.median())
print("Mean:", s.mean())
print("Standard Deviation:", s.std())

# Mencari nilai minimum dan maksimum
print("Minimum:", s.min())
print("Maximum:", s.max())

# Menghitung kuartil
q1 = s.quantile(0.25)
q2 = s.quantile(0.5)  # Setara dengan median
q3 = s.quantile(0.75)

print("Kuartil 1:", q1)
print("Kuartil 2:", q2)
print("Kuartil 3:", q3)

# Menampilkan statistik deskriptif secara lengkap sekaligus
print(s.describe())
```

```python
import matplotlib.pyplot as plt

# Mempersiapkan plot berdasarkan data series populasi
# Jenis plot yang dicontohkan: line (garis), bar (batang), box, dan pie (lingkaran)

# 1. Grafik Garis (Line Plot)
s.plot(kind='line')
plt.show()

# 2. Grafik Batang (Bar Chart)
s.plot(kind='bar')
plt.show()

# 3. Grafik Box Plot
s.plot(kind='box')
plt.show()

# 4. Grafik Lingkaran (Pie Chart)
s.plot(kind='pie')
plt.show()
```

```python
import pandas as pd

# 1. Baca file CSV menggunakan pemisah koma default (bukan titik koma)
df_pop = pd.read_csv('Angka Partisipasi Sekolah (APS) Menurut Provinsi dan Kelompok Umur, 2025.csv')

# Bersihkan nama kolom jika ada spasi atau koma yang terbawa
df_pop.columns = df_pop.columns.str.replace(r',+', '', regex=True).str.strip()

# 2. Urutkan berdasarkan kolom '38 Provinsi'
df_pop = df_pop.sort_values(by='38 Provinsi', ascending=True)

# 3. Reset indeks agar rapi dari 0 tanpa memunculkan kolom level_0
df_pop = df_pop.reset_index(drop=True)

# 4. Tampilkan data
df_pop.head()
```

```python
import pandas as pd

# 1. Baca data dan ubah nama kolom langsung
kolom = {'38 Provinsi': 'Provinsi', 'Unnamed: 1': '7-12', 'Unnamed: 2': '13-15', 'Unnamed: 3': '16-18', 'Unnamed: 4': '19-23'}
df = pd.read_csv('Angka Partisipasi Sekolah (APS) Menurut Provinsi dan Kelompok Umur, 2025.csv').rename(columns=kolom)

# 2. Filter baris bernilai valid, bersihkan teks non-provinsi (dengan na=False), urutkan, dan reset indeks
df_formatted = df[df['Provinsi'].notna() & ~df['Provinsi'].str.contains('Provinsi|Sumber|APS', case=False, na=False)].copy()
df_formatted = df_formatted.sort_values(by='Provinsi').reset_index(drop=True)

# Tampilkan hasil
display(df_formatted.head(10))
```

```python
# Pastikan kolom data bertipe numerik agar bisa dihitung statistiknya
for kol in ['7-12', '13-15', '16-18', '19-23']:
    df_formatted[kol] = pd.to_numeric(df_formatted[kol], errors='coerce')

# Menampilkan statistik deskriptif untuk data Angka Partisipasi Sekolah (APS)
print("Statistik Deskriptif Angka Partisipasi Sekolah (APS) 2025:")
display(df_formatted.describe())
```
